```python
###This program asks for the date of an event and displays it in understandable and appropriate form###

#Ask for required inputs
event = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

#Print the outputs
print("\nOutput:")
print("Event:", event.upper())
print("Date:", year, month, day)
print("Year:", year)
print("Month:", month)
print("Day:", day)
```
