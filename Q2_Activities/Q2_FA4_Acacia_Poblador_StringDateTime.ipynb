```python
###This program prints a students credentials using given data in clean and appropriate form###

#start
print('Please enter your credentials below')
print('\n')
#ask the user for the required inputs in string while eliminating trailing spaces
firstname = str(input('Enter First Name: ')).strip()
lastname  = str(input('Enter Last Name: ')).strip()
section = str(input('Enter Section: ')).strip()
favoritesubject = str(input('Enter Favorite Subject: ')).strip()
#combining the values of "firstname" and "lastname" to count the individual characters as a whole later in "fullname"
fullname = f'firstname, lastname'.strip()

#Outputs
print('\n')
print('Formatted Output:')
print('\n')

#Printing the full name in proper capitalization using .title()
print('Name:', firstname.title(), lastname.title())
#Printing the full name in all caps
print('Name Code:', firstname.upper(), lastname.upper())
#Printing the section in all caps
print('Section:', section.upper())
#Printing the favorite subject with appropriate capitalization
print('Favorite Subject:', favoritesubject.title())
#Print "Character Count:" then find the character count of "fullname" by putting it in len()
print('Character Count:', len(fullname))
```

```python
###This program asks for the date of an event and displays it in understandable and appropriate form###

import datetime

#Ask for required inputs
event = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

date = datetime.date(year, month, day)

#Print the outputs
print("\nFormatted Output:")
print("Event:", event.upper())
print("Date:", date.strftime("%B %d, %Y"))
print("Year:", year)
print("Month:", date.strftime("%B"))
print("Day:", day)
print("Weekday:", date.strftime("%A"))
```
