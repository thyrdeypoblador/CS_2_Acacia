```python
#set values

secret = 17
attempts = 0

#iteration using while
while attempts < 5:
    guess = int(input("Enter your guess: "))

    if guess < 0:
        continue

    attempts += 1
#codecracked
    if guess == secret:
        print("You won!")
        break

#condition
else:
    print("Game Over!")
```
