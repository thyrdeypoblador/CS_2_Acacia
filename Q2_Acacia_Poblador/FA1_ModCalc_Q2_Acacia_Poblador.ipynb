```python
def add(num1, num2):
  return num1 + num2

def subtract(num1, num2):
  return num1 - num2

def multiply(num1, num2):
  return num1 * num2

def divide(num1, num2):
  return num1 / num2

def main():
 num1 = float(input("Enter first number value: "))
 num2 = float(input("Enter second number value: "))

 print("Choices: [1: Addition] [2: Subtraction] [3: Multiplication] [4: Division]")
 choice = int(input("Select an operation [1/2/3/4]: "))

 if choice == 1:
   result = add(num1, num2)

 elif choice == 2:
   result = subtract(num1, num2)

 elif choice == 3:
   result = multiply(num1, num2)

 elif choice == 4:
   result = divide(num1, num2)

 else:
   print("Please only enter a number from the provided options [1/2/3/4]")

 print("Result: ", result)

main()
```

#What functions did you create in your calculator program?
#Ans: Functions that can calculate an operation including: addition, subtraction, mulitiplication, and division using two values.

#What parameters did your functions use?
#Ans: The parameters my functions used were the "num1" and "num2" parameters representing the two stored inputs the user has given globally to calculate with an operation.

#What arguments were passed when the functions were called?
#Ans: The arguments passed to the functions were the num1 and num2 variables. These variables held the floating-point numbers entered by the user during the execution of the main() function.

#How did your program use the returned value?
#Ans: The program caught the returned value from the called function and assigned it to a variable named result. It then used this variable to print the final calculation output to the screen (print("Result: ", result)).
